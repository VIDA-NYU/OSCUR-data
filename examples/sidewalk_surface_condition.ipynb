```python
import pandas as pd
import folium
from folium.plugins import MarkerCluster
from pathlib import Path

# ── 1. Pick the merged file ───────────────────────────────
merged_path = Path("../data/sidewalk_surface_condition/sidewalk_surface_full_merged.csv")

df = pd.read_csv(merged_path, low_memory=False)

# ── 2. Sample 5 000 rows for speed ────────────────────────
sample = df.sample(n=5000, random_state=1)

# ── 3. Color map for each source ──────────────────────────
palette = {
    "311_complaint": "orange",
    "violation":      "red",
    "tree_root_damage": "purple",
}

# Fallback color if other sources appear
default_color = "blue"

# ── 4. Build Folium map ───────────────────────────────────
m = folium.Map(
    location=[sample["lat"].mean(), sample["lon"].mean()],
    zoom_start=11,
    tiles="CartoDB Positron"
)

cluster = MarkerCluster().add_to(m)

for _, row in sample.iterrows():
    folium.CircleMarker(
        location=(row["lat"], row["lon"]),
        radius=4,
        weight=0,
        fill=True,
        fill_opacity=0.8,
        color=palette.get(row["merged_source"], default_color),
        popup=folium.Popup(
            f"<b>Source:</b> {row['merged_source']}<br>"
            f"<b>Descriptor:</b> {row.get('Descriptor', row.get('ViolationID', ''))}<br>"
            f"<b>Street:</b> {row.get('Street Name', row.get('street', ''))}",
            max_width=250
        )
    ).add_to(cluster)

m
```
