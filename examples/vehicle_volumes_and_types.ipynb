```python
import pandas as pd
import numpy as np
import folium

# ---- paths ----
CSV = "../data/vehicle_volumes_and_types/vehicle_counts_merged.csv"

# ---- load & clean ----
df = pd.read_csv(CSV, low_memory=False)

# ensure numeric lon/lat
df["lon"] = pd.to_numeric(df.get("lon"), errors="coerce")
df["lat"] = pd.to_numeric(df.get("lat"), errors="coerce")

# drop missing and clip to NYC-ish bbox
NYC_BBOX = {"min_lon": -74.30, "max_lon": -73.60, "min_lat": 40.45, "max_lat": 40.95}
df = df.dropna(subset=["lon", "lat"])
df = df[
    (df["lon"].between(NYC_BBOX["min_lon"], NYC_BBOX["max_lon"])) &
    (df["lat"].between(NYC_BBOX["min_lat"], NYC_BBOX["max_lat"]))
].copy()

# sample up to 3000 rows
SAMPLE_N = 3000
sample_df = df.sample(n=min(SAMPLE_N, len(df)), random_state=42)

# ---- map setup ----
center = [sample_df["lat"].mean(), sample_df["lon"].mean()]
m = folium.Map(location=center, zoom_start=11, tiles="cartodbpositron")

# layer groups by source
layers = {
    "classification": folium.FeatureGroup(name="Vehicle Classification (SegmentID→PHYSICALID)", show=True),
    "atr": folium.FeatureGroup(name="ATR Counts (nearest centerline)", show=True),
}
for layer in layers.values():
    m.add_child(layer)

# color by source
color_map = {"classification": "red", "atr": "green"}

# ---- add points ----
for _, row in sample_df.iterrows():
    src = row.get("source", "unknown")
    color = color_map.get(src, "blue")
    popup_bits = []
    for k in ["SegmentID", "PHYSICALID", "road_name", "from_street", "to_street", "direction", "date_time", "total_vehicles"]:
        if k in row and pd.notna(row[k]):
            popup_bits.append(f"<b>{k}</b>: {row[k]}")
    popup_html = folium.Popup(html="<br>".join(popup_bits) or src, max_width=350)

    folium.CircleMarker(
        location=[row["lat"], row["lon"]],
        radius=2,
        color=color,
        weight=1,
        fill=True,
        fill_opacity=0.6,
        popup=popup_html
    ).add_to(layers.get(src, m))

folium.LayerControl(collapsed=False).add_to(m)
m
```
