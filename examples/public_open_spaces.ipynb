```python
import pandas as pd
import geopandas as gpd
import folium
from shapely import wkt

# Load and convert
df = pd.read_csv("../data/public_open_spaces/public_open_spaces_final.csv")
df["geometry"] = df["geometry_wkt"].apply(wkt.loads)
gdf = gpd.GeoDataFrame(df, geometry="geometry", crs="EPSG:4326")

# Convert all geometries to centroids
gdf["geometry"] = gdf.geometry.centroid

# Sample 1000
sample = gdf.sample(n=1000, random_state=42)

# Create map centered on mean
center = [sample.geometry.y.mean(), sample.geometry.x.mean()]
m = folium.Map(location=center, zoom_start=12)

# Add markers
for _, row in sample.iterrows():
    folium.CircleMarker(
        location=[row.geometry.y, row.geometry.x],
        radius=3,
        color="blue",
        fill=True,
        fill_opacity=0.6,
        popup=row.get("source", "N/A")
    ).add_to(m)

m
```
