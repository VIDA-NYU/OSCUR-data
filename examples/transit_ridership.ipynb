```python
import pandas as pd
import folium

# Load your CSV (adjust path if needed)
df = pd.read_csv("../data/transit_ridership/mta_subway_hourly_ridership.csv")

# Sample 3000 rows
sample_df = df.sample(n=3000, random_state=42)

# Create a folium map centered on NYC
m = folium.Map(location=[40.7128, -74.0060], zoom_start=11)

# Add points to map
for _, row in sample_df.iterrows():
    folium.CircleMarker(
        location=[row["latitude"], row["longitude"]],
        radius=2,
        color='blue',
        fill=True,
        fill_opacity=0.6,
    ).add_to(m)

# Display in notebook
m
```
