```python
import pandas as pd
import geopandas as gpd
import folium
from shapely import wkt

# ---- Load merged file ----
path = "../data/roadway_features/roadway_features_merged.csv"
df = pd.read_csv(path, low_memory=False)

# ---- Find geometry column robustly ----
candidate_wkts = ["geometry_wkt", "geom_wkt", "centerline_wkt", "wkt"]
wkt_col = next((c for c in candidate_wkts if c in df.columns), None)

if wkt_col:
    df["geometry"] = df[wkt_col].apply(wkt.loads)
    gdf = gpd.GeoDataFrame(df, geometry="geometry", crs="EPSG:4326")
    # centerlines are lines; show centroids for markers
    gdf["geometry"] = gdf.geometry.centroid
elif {"lon", "lat"}.issubset(df.columns):
    # Fallback if you exported lon/lat instead of WKT
    gdf = gpd.GeoDataFrame(
        df,
        geometry=gpd.points_from_xy(df["lon"], df["lat"]),
        crs="EPSG:4326",
    )
else:
    raise KeyError(
        "No geometry found. Looked for WKT columns "
        f"{candidate_wkts} or lon/lat. Available columns:\n{list(df.columns)}"
    )

# ---- Sample for speed ----
sample = gdf.sample(n=1000, random_state=42) if len(gdf) > 1000 else gdf.copy()

# ---- Center the map on the sample ----
center = [sample.geometry.y.mean(), sample.geometry.x.mean()]
m = folium.Map(location=center, zoom_start=12)

# ---- Build popups using your prefixed columns (if present) ----
def yn(val):
    return "yes" if pd.notna(val) else "no"

for _, row in sample.iterrows():
    popup_text = (
        f"PHYSICALID: {row.get('PHYSICALID', 'N/A')}<br>"
        f"Bus lane join: {yn(row.get('bus_SegmentID'))}<br>"
        f"Corridor join: {yn(row.get('cor_ProjectName'))}<br>"
        f"Intersection join: {yn(row.get('int_mid_lat')) or yn(row.get('int_the_geom'))}<br>"
        f"Blockface join: {yn(row.get('blk_mid_lat')) or yn(row.get('blk_block_id'))}"
    )
    folium.CircleMarker(
        location=[row.geometry.y, row.geometry.x],
        radius=3,
        color="blue",
        fill=True,
        fill_opacity=0.6,
        popup=folium.Popup(popup_text, max_width=320),
    ).add_to(m)

m
```
