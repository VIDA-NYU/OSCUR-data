```python
import pandas as pd
import geopandas as gpd
import folium
import json
from shapely import wkt
from shapely.geometry import shape as shp_shape

CSV_PATH = "../data/environmental_justice/dac_2023_nyc_final.csv"  # your NYC-only file
GEOM_CANDIDATES = ["the_geom", "geometry", "shape", "geometry_wkt"]  # possible geometry columns

df = pd.read_csv(CSV_PATH, low_memory=False)

# find geometry column
geom_col = next((c for c in GEOM_CANDIDATES if c in df.columns), None)
if geom_col is None:
    raise ValueError(f"No geometry column found. Expected one of: {', '.join(GEOM_CANDIDATES)}")

def parse_geom(s):
    if pd.isna(s): return None
    s = str(s).strip()
    if not s: return None
    # try WKT
    try:
        g = wkt.loads(s)
        return g
    except Exception:
        pass
    # try (feature) GeoJSON
    try:
        gj = json.loads(s)
        geom_obj = gj.get("geometry", gj)  # feature or bare geometry
        return shp_shape(geom_obj)
    except Exception:
        return None

geoms = df[geom_col].apply(parse_geom)
gdf = gpd.GeoDataFrame(df, geometry=geoms, crs="EPSG:4326").dropna(subset=["geometry"])

# keep only designated DACs (case-insensitive)
flag_col = "DAC_Designation"  # change if needed
if flag_col in gdf.columns:
    gdf = gdf[gdf[flag_col].astype(str).str.lower().eq("designated as dac")].copy()

# map
minx, miny, maxx, maxy = gdf.total_bounds
m = folium.Map(location=[(miny+maxy)/2, (minx+maxx)/2], zoom_start=12, tiles="cartodbpositron")

folium.GeoJson(
    gdf.__geo_interface__,
    style_function=lambda _ : {"color":"#2c7fb8","weight":0.8,"fillOpacity":0.35},
    highlight_function=lambda _ : {"color":"#000","weight":2.0,"fillOpacity":0.55},
).add_to(m)

m.fit_bounds([[miny, minx], [maxy, maxx]])
m
```
