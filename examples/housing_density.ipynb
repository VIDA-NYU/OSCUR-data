```python
import pandas as pd
import geopandas as gpd
import folium
import json
from shapely import wkt
import numpy as np
import branca

# ── CONFIG ─────────────────────────────────────────────────────────────
CSV_PATH = "../data/housing_density/housing_by_cdta_2020.csv"  # adjust if needed
GEOM_COL = "the_geom"      # WKT MULTIPOLYGON from the housing DB
NAME_COL = "cdtaname20"    # CDTA name
UNITS_COL = "cenunits20"   # total Class A units (2020 baseline)

# ── LOAD & GEOMETRY ───────────────────────────────────────────────────
df = pd.read_csv(CSV_PATH, low_memory=False)

# Parse WKT geometry
geom = df[GEOM_COL].apply(lambda s: wkt.loads(s) if isinstance(s, str) and s.strip() else None)
gdf = gpd.GeoDataFrame(df, geometry=geom, crs="EPSG:4326").dropna(subset=["geometry"])

# ── AREA & DENSITY (units/acre & units/sqmi) ──────────────────────────
# Use a local projected CRS for accurate area (NY Long Island feet)
gdf_proj = gdf.to_crs(2263)  # EPSG:2263 (ft)
area_sqft = gdf_proj.geometry.area
area_acres = area_sqft / 43560.0
area_sqmi  = area_acres / 640.0

gdf["area_acres"] = area_acres.values
gdf["area_sqmi"]  = area_sqmi.values

# Avoid divide-by-zero
gdf["units_per_acre"] = gdf[UNITS_COL] / gdf["area_acres"].replace({0: np.nan})
gdf["units_per_sqmi"] = gdf[UNITS_COL] / gdf["area_sqmi"].replace({0: np.nan})

# ── FOLIUM CHOROPLETH (by units per acre) ─────────────────────────────
min_val = float(gdf["units_per_acre"].min())
max_val = float(gdf["units_per_acre"].max())
cmap = branca.colormap.linear.YlOrRd_09.scale(min_val, max_val)
cmap.caption = "Housing Density (units per acre)"

def style_fn(feature):
    # feature["id"] is not set; values are in properties
    v = feature["properties"].get("units_per_acre", None)
    color = "#cccccc" if v is None or pd.isna(v) else cmap(v)
    return {"fillColor": color, "color": "#555", "weight": 0.7, "fillOpacity": 0.7}

# Center map on dataset extent
minx, miny, maxx, maxy = gdf.total_bounds
center = [(miny + maxy) / 2.0, (minx + maxx) / 2.0]
m = folium.Map(location=center, zoom_start=11, tiles="cartodbpositron")

# Choose a few popup fields if present
popup_fields = [
    NAME_COL, UNITS_COL, "area_acres", "units_per_acre", "units_per_sqmi"
]
popup_aliases = ["CDTA", "Total Units (2020)", "Area (acres)", "Units / Acre", "Units / Sq Mi"]

folium.GeoJson(
    data=gdf[["geometry"] + popup_fields].to_json(),
    style_function=style_fn,
    highlight_function=lambda _: {"weight": 2, "color": "#000", "fillOpacity": 0.8},
    popup=folium.GeoJsonPopup(fields=popup_fields, aliases=[f"{a}:" for a in popup_aliases], localize=True),
    name="Housing Density (units/acre)"
).add_to(m)

cmap.add_to(m)
folium.LayerControl().add_to(m)

# Fit to bounds
m.fit_bounds([[miny, minx], [maxy, maxx]])

m
```
