```python
import pandas as pd
import geopandas as gpd
import folium

# Load datasets
parks = pd.read_csv("../data/environment/parks.csv")
open_space = pd.read_csv("../data/environment/open_space.csv")

# Convert to GeoDataFrame
if 'the_geom' in parks.columns:
    parks = parks.assign(geometry=gpd.GeoSeries.from_wkt(parks['the_geom']))
    parks = gpd.GeoDataFrame(parks, geometry='geometry', crs="EPSG:4326")

if 'the_geom' in open_space.columns:
    open_space = open_space.assign(geometry=gpd.GeoSeries.from_wkt(open_space['the_geom']))
    open_space = gpd.GeoDataFrame(open_space, geometry='geometry', crs="EPSG:4326")

# Create folium map centered on NYC
m = folium.Map(location=[40.7128, -74.0060], zoom_start=11)

# Add layers
folium.GeoJson(parks.sample(200), name="Parks Sample").add_to(m)
folium.GeoJson(open_space.sample(200), name="Open Space Sample").add_to(m)

folium.LayerControl().add_to(m)

m
```
